```python
# 7.1 Print i as lomg as i is less than 6:
i=1
while i<=5:
  print (i)
  i+=1
```

```python
# 7.2 Print numbers from 1 to N
n= int(input("Enter N: "))
i=1
while i<=n:
  print(i)
  i+=1
```

```python
# 7.3 Factorial of a number
n= int(input("Enter a number: "))
fact=1
while n>0:
  fact *= n
  n-=1
print("Factorial=", fact)
```

```python
# 7.4 Fibonacci Series using while loop
n= int(input("Enter number of terms: "))
a=0
b=1
i=1
while i<=n:
  print(a, end= " ")
  c=a + b
  a=b
  b=c
  i +=1
```

```python
# Using limit in fibonacci series
limit=int(input("Enter limit: "))
a=0
b=1
while a<=limit:
  print(a, end=" ")
  a, b=b, a+b
```

```python
# 7.5 Reverse a number
num=int(input("Enter a number: "))
rev=0
while num>0:
  digit=num % 10
  rev= rev * 10 + digit
  num //=10
print("Reversed number =",rev)
```

```python
# 7.6 Check Palindrome number/string
num=int(input("Enter a number: "))
temp = num
rev = 0
while num > 0:
    rev= rev*10 + num % 10
    num //=10
if temp == rev:
   print("Pallindrome")
else:
   print("Not Pallindrome")
```

```python
s=input("Enter the string: ")
i,j = 0, len(s)-1
is_palindrome = True
while i < j:
  if s[i] !=s[j]:
    is_palindrome= False
    break
  i+=1
  j-=1
if is_palindrome:
  print("Yes")
else:
  print("No")
```

```python
# Check Palindrome using if else: (works with both numbers and letters) and numbers and letters are stored as string.
st= input("Enter a string:")
rev=st[::-1]
if st==rev:
  print("Palindrome")
else:
  print("Not a Palindrome")
```

```python
#7.7 Count digits in a numbe number
num=int(input("Enter a number: "))
count=0
while num>0:
  count+=1
  num //=10
print("Number of digits =",count)
```

The Break Statement- With the break statement we can stop the loop even if the while condition is True.

```python
 #7.9 Search an element in a list
 nums=[10,20,30,40,50]
 key= int(input("Enter element to search:"))
 i=0
 while i<len(nums):
   if nums[i]==key:
     print("Element found in index",i+1)
     break
   i+=1
 else:
     print("Element not found")
```

```python
#7.8 Exit the loop when i is 3:
i=1
while i < 6:
  print(i)
  if i==3:
    break
  i +=1
```

The Continue Statement- Skips the current iteration and continues with the next.

```python
#7.10 To print only odd numbers(1 to 10)
i=0
while i<10:
  i+=1
  if i%2==0:
    continue
  print(i)
```
